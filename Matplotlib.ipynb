# Bar Graph

```python
x=[1,2,3,4,5]
y=["Monday","Tuesday","Wednesday","Thursday","Friday"]
```

```python
plt.bar(x,y)
plt.title("day wise sales")
plt.show()
```

```python
x=[1,2,5,3,6,2]
plt.pie(x  , autopct ="%1.1f%%")
plt.show()
```
